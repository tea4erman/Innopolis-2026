### Метод loc (location) для выбора столбцов и строк по условию

```python
# Метод loc пишется через точку после датафрейма
# В отличие от всех остальных методов (кроме iloc) его аргументы пишутся в КВАДРАТНЫХ скобках
# Первая позиция в loc - это название строки / строк (здесь можно указать логическое выражение, чтобы выбрать нужные строки)
# Вторая позиция в loc - это название столбца
prices_df.loc['Январь', 'Яблоки']
```

```python
# Если нужно выбрать столбец целиком, то на месте строк ставится :
# Если нужно выбрать одну строку, то на месте столбцов ставится :
prices_df.loc[:, 'Яблоки']
```

```python
# Можно вызвать несколько столбцов, указав их в списке
prices_df.loc[:, ['Яблоки', 'Апельсины']]
```

```python
# Можно вызвать отдельно строку, указав ее имя в .loc[]
prices_df.loc['Январь']
```

```python
# Равносильно операции выше, но с указанием : на месте столбцов
prices_df.loc['Январь', :]
```

```python
# Можно вызвать несколько строк и столбцов (не обязательно смежных)
prices_df.loc[['Январь', 'Март'], ['Яблоки', 'Апельсины']]
```

```python
# Можно вызвать столбцы, указав их в квадратных скобках (способ 1), а строки вызвать с помощью loc
prices_df[['Яблоки', 'Апельсины']].loc[['Январь','Март']]
```

## Создание новых столбцов

### Создание столбца напрямую

```python
# Продолжительность аренды (разность двух datetime объектов создает объект типа timedelta)
rentals["rental_length"] = rentals["return_date"] - rentals["rental_date"]
rentals.head()
```

```python
# внутри метода assign указываем (имя_переменной = lambda x: формула расчета переменной с использованием x в качестве заменителя имени датафрейма)
rentals = rentals.assign(rental_length2 = lambda x: x["return_date"] - x["rental_date"])
rentals.head()
```

## Объединение таблиц (аналог JOIN)

```python
# Простой джойн, где pandas самостоятельно определяет, по какой колонке и как его сделать
# Плохая практика, т.е. есть шансы ошибиться
pd.merge(rentals[["rental_id", "customer_id"]], customers)
```

## Примеры объединения таблиц

```python
a = pd.DataFrame({"id": [1,2,3,4], "age": [30, 40, 35, 45]})
b = pd.DataFrame({"id": [1,2,3,5], "name" : ["Ivan", "Olga", "Boris", "Elena"]})
```

```python
# Left join
pd.merge(left = a, right = b, how = "left", left_on = "id", right_on = "id")
```

```python
# Right join
pd.merge(left = a, right = b, how = "right", left_on = "id", right_on = "id")
```

```python
# Inner join
pd.merge(left = a, right = b, how = "inner", left_on = "id", right_on = "id")
```

```python
# Full join
pd.merge(left = a, right = b, how = "outer", left_on = "id", right_on = "id")
```

```python
# Cross join
c = pd.DataFrame({"client":[1,2,3,4], "address":["a", "b", "c", "d"]})
d = pd.DataFrame({"courier":[1,2,3,4,5], "address":["c","d","a","b","e"]})
(
    pd.merge(left = c, right = d, how = "cross", suffixes = ("_client", "_courier"))
    .query("address_client == address_courier")
)
```

```python
# В Pandas версии 3.0 и выше
# Left antijoin
# pd.merge(left = a, right = b, how = "left_anti", left_on = "id", right_on = "id")
```

### Расчет агрегатов по группам

```python
# Расчет количества прокатов на одного клиента
# SELECT customer_id, COUNT(rental_id) FROM rental GROUP BY customer_id
# Вариант 1: без использования команды AGG
rentals.groupby("customer_id")["rental_id"].count()
```

```python
# Расчет количества прокатов на одного клиента
# Вариант 2: используя команду AGG
rentals.groupby("customer_id").agg({"rental_id":"count"})
```

```python
# Расчет среднего времени проката на каждого клиента
rentals["fact_rental_duration"] = rentals["return_date"] - rentals["rental_date"]
rentals.groupby("customer_id").agg({"fact_rental_duration":"mean"})
```

## Groupby / Transform (аналог оконных функций)

Рассчитаем процентные изменения выручки от каждого клиента по месяцам

```python
# 1: трансформируем payment date из строки в datetime
# 2: трансформируем datetime в period
# 3: трансформируем period в timestamp
payment["payment_month"] = pd.to_datetime(payment["payment_date"]).dt.to_period("M").dt.to_timestamp()
monthly_payments = payment.groupby(["customer_id", "payment_month"]).agg({"amount" : "sum"}).reset_index(drop = False)
sorted_payments = monthly_payments.sort_values(by = ["customer_id", "payment_month"])
sorted_payments["amount_lag"] = sorted_payments.groupby("customer_id")["amount"].shift()
sorted_payments["pct_change"] = sorted_payments["amount"] / sorted_payments["amount_lag"] - 1
sorted_payments
```

### Пропуски можно заполнить

Если пропуск в ID, то такие записи лучше удалять. Вы все равно не сможете понять, кому принадлежит эта запись.

Если пропуск в дате, то надо понимать, почему он появился. Если событие еще не случилось, то можно заполнить пропуски какой-то будущей датой ("2999-01-01")

Пример: возврат аренды DVD. Те прокаты, которые еще не вернули, имеют пропуски в столбце return_date

Если пропуск в какой-то качественной характеристике (название фильма, адрес, телефон и т.п.), можно заполнить его какой-то специальной записью (например, -1 или NA, НД и т.п.)

Если пропуск в числовой переменной, то можно заполнять средней, медианой и т.п.

```python
import pandas as pd

customers = pd.read_csv(r"C:\Courses\Innopolis\2026\ИИ Инженер\Python Data\customer.csv")
customers["customer_name"] = customers["first_name"] + " " + customers["last_name"]
customers = customers.drop(["first_name", "last_name", "activebool", "create_date", "last_update", "active"], axis = 1)

rentals = pd.read_csv(r"C:\Courses\Innopolis\2026\ИИ Инженер\Python Data\rental.csv")
rentals["return_date"] = pd.to_datetime(rentals["return_date"])
rentals["rental_date"] = pd.to_datetime(rentals["rental_date"])
rentals["fact_rental_duration"] = rentals["return_date"] - rentals["rental_date"]
rentals["flag_returned"] = np.where(rentals["return_date"].isna(), 0, 1)
rentals = rentals.loc[:, ["rental_id", "rental_date", "return_date", "fact_rental_duration", "flag_returned", "inventory_id", "customer_id", "staff_id"]]

inventory = pd.read_csv(r"C:\Courses\Innopolis\2026\ИИ Инженер\Python Data\inventory.csv")
inventory = inventory.drop("last_update", axis = 1)

film = pd.read_csv(r"C:\Courses\Innopolis\2026\ИИ Инженер\Python Data\film.csv")
film = film.loc[:, ["film_id", "rental_duration", "title", "rental_rate"]]
film["rental_duration"] = pd.to_timedelta(film['rental_duration'], unit='D')
film = film.rename(columns = {"rental_duration" : "normative_rental_duration", "title" : "film_name"})

payment = pd.read_csv(r"C:\Courses\Innopolis\2026\ИИ Инженер\Python Data\payment.csv")
payment = payment.drop(["customer_id", "staff_id"], axis = 1)

result_table = pd.merge(left = rentals, right = customers, how = "left", left_on = "customer_id", right_on = "customer_id")
result_table = pd.merge(left = result_table, right = inventory, how = "left", left_on = "inventory_id", right_on = "inventory_id")
result_table = pd.merge(left = result_table, right = film, how = "left", left_on = "film_id", right_on = "film_id")
result_table = pd.merge(left = result_table, right = payment, how = "left", left_on = "rental_id", right_on = "rental_id")
result_table["rental_overdue"] = result_table["fact_rental_duration"] - result_table["normative_rental_duration"]
result_table = result_table.sort_values(by = "rental_date")

result_table.head()
```
