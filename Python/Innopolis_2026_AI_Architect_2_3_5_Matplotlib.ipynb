## Изменение параметров графика

```python
# Изменим цвет линии
x = np.arange(-10,11)
y = x**2
plt.plot(x, y, color = 'mediumvioletred')
plt.show()
```

```python
# Подпишем линию для будущей легенды
plt.figure(figsize = (15,4))
x = np.arange(-10,11,0.01)
y = np.sin(x)
plt.plot(x, y, label = 'Синус', color = 'red')
plt.show()
```

```python
# Рисуем линию светло-голубого цвета
x = [1,2,3,4,5]
y = [10,20,30,40,50]
plt.plot(x, y, color = 'deepskyblue')
plt.show()
```

```python
# Изменим стиль линии
x = [1,2,3,4,5]
y = [10,20,30,40,50]
plt.plot(x, y, color = 'deepskyblue', linestyle = '--')
```

```python
# Изменим стиль линии через код (где начинается (длина штриха, длина пропуска, длина штриха, длина пропуска))
x = np.arange(-10,11,0.01)
y = np.sin(x)
plt.plot(x, y, label = 'Синус', color = 'red', linestyle = (0, (5,2,3,2)))
plt.show()
```

```python
# Добавим маркеры
x = np.arange(-10,11)
y = x**2
plt.plot(x, y, color = 'mediumvioletred', marker = 'X')
plt.show()
```

```python
# Изменим толщину линии
x = [1,2,3,4,5]
y = [10,20,30,40,50]
plt.plot(x, y, color = 'deepskyblue', linestyle = '-.', linewidth = 5)
plt.show()
```

```python
# Изменим стиль линии
x = np.arange(-10,11)
y = x**2
plt.plot(x, y, marker = 'x', linestyle = (0, (1, 10)))
plt.show()
```

```python
# Добавим заголовок с формулой LaTeX
# В заголовке меняем размер шрифта, меняем начертание (полужирное), цвет шрифта
# Меняем положение заголовка 
x = np.arange(-10,11)
y = x**2
plt.plot(x, y, marker = 'x', linestyle = (0, (1, 10)))
plt.title('График функции $y = x^2$', fontsize = 14, fontweight = 'bold', color = 'steelblue', loc = 'right')
plt.show()
```

```python
# Добавим заголовок с формулой LaTeX и подписями осей
x = np.arange(-10,11)
y = x**2
plt.plot(x, y, marker = 'x', linestyle = (0, (1, 10)))
plt.title('График функции $y = x^2$', fontsize = 10, fontweight = 'bold', color = 'steelblue', loc = 'right')
plt.xlabel('Значения переменной X')
plt.ylabel('Значения переменной Y')
plt.show()
```

```python
# Добавим заголовок с формулой LaTeX, подписями осей и сеткой
# В сетке указываем, что рисовать только основные линии
# В сетке прозрачность равна 0.2
x = np.arange(-10,11)
y = x**2
plt.plot(x, y, marker = 'x', linestyle = (0, (1, 10)))
plt.title('График функции $y = x^2$', fontsize = 10, fontweight = 'bold', color = 'steelblue', loc = 'right')
plt.xlabel('Значения переменной X')
plt.ylabel('Значения переменной Y')
plt.grid(which = 'major', axis = 'both', alpha = 0.2)
plt.show()
```

```python
# Изменим размер
x = np.arange(-10,11)
y = x**2
plt.figure(figsize = (15, 4))
plt.plot(x, y, marker = 'x', linestyle = (0, (1, 10)))
plt.title('График функции $y = x^2$', fontsize = 10, fontweight = 'bold', color = 'steelblue', loc = 'right')
plt.xlabel('Значения переменной X')
plt.ylabel('Значения переменной Y')
plt.grid(which = 'major', axis = 'both', alpha = 0.1)
plt.show()
```

```python
# Рисуем два графика на одном холсте
x = np.arange(-3,3.1, 0.1)
y1 = x**2
y2 = np.sqrt(np.abs(x))

plt.plot(x, y1)
plt.plot(x, y2)

plt.title('Графики функций $y = x^2$ и $y = \sqrt{x}$', fontsize = 10, fontweight = 'bold')
plt.xlabel('Значения переменной X')
plt.ylabel('Значения переменной Y')
plt.grid(which = 'major', axis = 'both', alpha = 0.1)
plt.show()
```

```python
# Выносим подписи функций в легенду
x = np.arange(-3,3.1, 0.1)
y1 = x**2
y2 = np.sqrt(np.abs(x))

plt.plot(x, y1, label = "$y = x^2$")
plt.plot(x, y2, label = "$y = \sqrt{x}$")

plt.title('Графики функций $y = x^2$ и $y = \sqrt{x}$', fontsize = 10, fontweight = 'bold')
plt.xlabel('Значения переменной X')
plt.ylabel('Значения переменной Y')
plt.grid(which = 'major', axis = 'both', alpha = 0.1)
plt.legend()
plt.show()
```

```python
# Настраиваем легенду
# Задаем локацию, заголовок легенды, рамку и цвет рамки и цвет заливки легенды
x = np.arange(-3,3.1, 0.1)
y1 = x**2
y2 = np.sqrt(np.abs(x))

plt.plot(x, y1)
plt.plot(x, y2)

plt.title('Графики функций $y = x^2$ и $y = \sqrt{x}$', fontsize = 10, fontweight = 'bold')
plt.xlabel('Значения переменной X')
plt.ylabel('Значения переменной Y')
plt.grid(which = 'major', axis = 'both', alpha = 0.1)

plt.legend(['$y = x^2$', '$y = \sqrt{x}$'], loc = 'upper left', title = 'Функции', frameon = True, facecolor = 'white', edgecolor = 'red')
plt.show()
```

```python
# Нанесение текста на график
x = np.arange(-10,11,0.01)
y = np.sin(x)
plt.plot(x, y, color = 'red', linestyle = (0, (3,1,1,1)))
plt.annotate("Это график функции синуса", xy = (-5, 1))
plt.show()
```

```python
# Наносим подписи значений на график
# Подписываем координаты точек на графике
x = np.arange(-10, 11, 1)
y1 = x**2
y2 = np.sqrt(np.abs(x))

plt.plot(x, y1)
plt.plot(x, y2)

plt.title('Графики функций $y = x^2$ и $y = \sqrt{x}$', fontsize = 10, fontweight = 'bold')
plt.xlabel('Значения переменной X')
plt.ylabel('Значения переменной Y')
plt.grid(which = 'major', axis = 'both', alpha = 0.1)

for i in range(len(x)):
    plt.annotate(text = f'({x[i]}, {y1[i]})'  , xy = (x[i],y1[i]))
plt.show()
```

```python
# Корректируем размеры осей графика с помощью xlim и ylim
x = np.arange(-10,11)
y = x**2
plt.plot(x, y, marker = 'x', linestyle = (0, (1, 10)))
plt.xlim(-50,50)
plt.ylim(0,100)
plt.title('График функции $y = x^2$', fontsize = 10, fontweight = 'bold', color = 'steelblue', loc = 'right')
plt.xlabel('Значения переменной X')
plt.ylabel('Значения переменной Y')
plt.grid(which = 'major', axis = 'both', alpha = 0.1)
```

```python
# Сохранение графика в файл
x = np.arange(-10,11)
y = x**2
plt.figure(figsize = (5,3))
plt.plot(x, y, marker = 'x')
plt.title('График функции $y = x^2$', fontsize = 16, fontweight = 'bold', color = 'steelblue', loc = 'right')
plt.xlabel('Значения переменной X', fontsize = 14)
plt.ylabel('Значения переменной Y', fontsize = 14)
plt.tight_layout()
plt.savefig(r"C:\Courses\Innopolis\2026\ИИ Инженер\Лекции\plot2.jpeg")
plt.show()
```

## Виды графиков в MatPlotLib

### Точечная диаграмма

```python
x = [1,2,3,4,5]
y = [10,20,30,40,50]
plt.scatter(x, y)
plt.show()
```

### Круговая диаграмма

```python
sizes = [35, 15, 5, 45]
labels = ['A', 'B', 'AB', 'O']
plt.pie(sizes, labels = labels, autopct = '%0.2f%%', startangle = 120)
plt.show()
```

### Столбчатая диаграмма

```python
sizes = [35, 15, 5, 45]
labels = ['A', 'B', 'AB', 'O']
plt.bar(x = labels, height = sizes,  bottom = 0)
for i in range(len(sizes)):
    plt.annotate(text = f'{sizes[i]}%', xy = (labels[i], sizes[i]+1))
plt.tight_layout()
plt.ylim(0, 50)
plt.show()
```

```python
sizes = [35, 45]
labels = ['A',  'O']
plt.bar(x = labels, height = sizes)
plt.ylim(30, 50) # Режем границы оси У
plt.show()

# Якобы двукратная разница между категориями
# На самом деле - разница 35 и 45 не двукратная!
```

### Гистограмма

```python
x = np.random.randn(1000)
plt.hist(x, bins = 10)
plt.show()
```

```python
# Нормализуем высоту столбиков, чтобы они суммировались к 1
plt.hist(x, bins = 10, density = True)
plt.show()
```

```python
# Кумулятивная гистограмма
plt.hist(x, bins = 10, density = True, cumulative = True)
plt.show()
```

## Расположение нескольких графиков на одной полотне

```python
sizes = [35, 15, 5, 45]
labels = ['A', 'B', 'AB', 'O']

fig, ax = plt.subplots(1, 2, figsize = (10, 5))
ax[0].pie(sizes, labels = labels, autopct = '%0.2f%%', startangle = 120)
ax[1].bar(labels, sizes)
for i in range(len(sizes)):
    ax[1].annotate(text = f'{sizes[i]}%', xy = (labels[i], sizes[i]+1))
plt.show()
```

```python
# При расположении графиков на одном полотне, настройки заголовков и осей происходят иначе!
fig, ax = plt.subplots(1, 2, figsize = (10, 5))
ax[0].pie(sizes, labels = labels, autopct = '%0.2f%%', startangle = 180)
ax[0].set_title('Круговая')
ax[1].bar(labels, sizes)
ax[1].set_title('Столбчатая')
ax[1].set_xlabel('Группы крови')
plt.suptitle('Комбинированный график', fontsize = 16)
plt.tight_layout()
plt.show()
```
