```python
from selenium import webdriver
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
import time 
```

```python
def open_browser(url_address):
    driver = webdriver.Chrome()
    driver.get(url_address)
    time.sleep(5)
    return driver
    
def enter_address(driver, delay, address):
    
    elem_search_string = WebDriverWait(driver, delay) \
        .until(EC.presence_of_element_located(
        (By.XPATH, "//input[@class='input__control _bold']")))

    elem_search_string.send_keys(address)

    elem_search_string.send_keys(Keys.ENTER)

def get_coordinates(driver, delay):
    try:
        elem_search_coords = WebDriverWait(driver, delay) \
            .until(EC.presence_of_element_located(
            (By.XPATH, "//div[@class='toponym-card-title-view__coords-badge']")))
        # Запись в переменную координат адреса
        coord = elem_search_coords.text
        return coord#.replace(" ","").split(",")
    except:
        return 'no coordinates'

def select_first_option(driver, delay):
    try:
        elem_first_list = WebDriverWait(driver, delay) \
            .until(EC.presence_of_element_located(
            (By.XPATH, "//div[@class='search-snippet-view__body _type_toponym']")))
        elem_first_list.click()
    except:
        pass
    
def clear_search_string(driver, delay):
    try:
        elem_clear = WebDriverWait(driver, delay) \
            .until(EC.presence_of_element_located(
            (By.XPATH, "//a[@class='small-search-form-view__pin']")))
    except:
        elem_clear = WebDriverWait(driver, delay) \
            .until(EC.presence_of_element_located(
            (By.XPATH, "//div[@class='small-search-form-view__icon _type_close']")))
    elem_clear.click()
```

```python
url_address = 'https://yandex.ru/maps'
driver = open_browser(url_address)
    
addresses = [
    "г. Москва, ул. Звенигородская 2-я, д. 13, стр. 15",
    "Татарстан, г. Иннополис, ул. Университетская, д. 1"
]

for address in addresses:
    enter_address(driver, delay = 10, address = address)
    select_first_option(driver, delay = 10)
    coord = get_coordinates(driver, delay = 10)
    clear_search_string(driver, delay = 2)
    print(address, coord)

driver.quit()
```
