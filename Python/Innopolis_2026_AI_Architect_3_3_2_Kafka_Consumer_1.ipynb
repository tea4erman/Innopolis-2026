# Consumer 1

```python
from kafka import KafkaConsumer
import json
```

```python
consumer = KafkaConsumer(
    "purchases",

    bootstrap_servers="localhost:9092",

    # Преобразуем JSON обратно в Python-словарь
    value_deserializer=lambda x: json.loads(x.decode("utf-8")),

    # Название группы потребителей
    group_id="sales_analytics",

    # Если группа ещё не читала сообщения, начинаем с самого первого
    auto_offset_reset="earliest"
)
```

```python
# Потребитель - это приложение, которое считает количество заказов и суммарную выручку
total_revenue = 0
orders_count = 0

print("Consumer started. Waiting for events...")

try:
    for message in consumer:

        event = message.value

        orders_count += 1
        total_revenue += event["amount"]

        print("\nNew purchase received:")
        print(f"Customer: {event['customer_id']} | Product: {event['product_id']} | Amount: {event['amount']}")

        print("\nCurrent analytics:")
        print(f"Orders: {orders_count} | Revenue: {total_revenue}")

except KeyboardInterrupt:
    print("Consumer stopped")

finally:
    consumer.close()
```
