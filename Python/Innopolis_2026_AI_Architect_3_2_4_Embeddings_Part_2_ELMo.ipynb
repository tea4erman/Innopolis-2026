```python
import numpy as np

from sklearn.metrics.pairwise import cosine_similarity

import tensorflow as tf
import tensorflow_hub as hub
```

```python
elmo = hub.load("https://tfhub.dev/google/elmo/3")
```

```python
elmo
```

```python
docs =  [
    "cats chase mice",
    "dogs chase cats",
    "mice love cheese",
    "rats love cheese",
    "dogs bark at cats loudly",
    "I withdrew money from a bank",
    "I sat on a river bank"
  ]
```

```python
outputs = elmo.signatures["default"](tf.constant(docs))
```

```python
outputs.keys()
```

```python
embeddings = outputs["elmo"]
```

```python
print(f"Размерность эмбеддингов: {embeddings.shape}")
```

```python
print(f"Пример эмбеддинга для первого предложения: {embeddings[0][:10]}...")
```

```python
print(f"Эмбеддинг для слова dogs из второго предложения:", embeddings[1, 0])
```

```python
print(f"Эмбеддинг для слова dogs из пятого предложения:", embeddings[4, 0])
```

```python
cos_sim = cosine_similarity(embeddings[1, 0].numpy().reshape(1, -1), embeddings[4, 0].numpy().reshape(1, -1))
print("ELMO Cosine Similarities для слов dogs:", np.round(cos_sim, 2))
```

```python
cos_sim = cosine_similarity(embeddings[5, -1].numpy().reshape(1, -1), embeddings[6, -1].numpy().reshape(1, -1))
print("ELMO Cosine Similarities для слов bank:", np.round(cos_sim, 2))
```
