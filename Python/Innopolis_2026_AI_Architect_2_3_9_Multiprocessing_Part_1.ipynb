# Три уровня кода

## Исходный код

```python
x = 10
y = 20
z = x + y
print(z)
```

## Байткод

```python
import dis

def add():
    x = 10
    y = 20
    return x + y

dis.dis(add)
```

### бинарное представление байткода CPython

```python
print(add.__code__.co_code)
```

### 16-ричное представление байткода CPython

```python
print(add.__code__.co_code.hex())
```
