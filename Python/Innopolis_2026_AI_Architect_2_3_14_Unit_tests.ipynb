**Виды тестирования**

1) Unit-тест: Проверяет одну функцию/метод в изоляции
2) Integration-тест: Проверяет взаимодействие нескольких модулей
3) End-to-end (E2E) тест: Проверяет	весь сценарий целиком, как у пользователя
4) Regression-тест: Проверяет, что то, что раньше работало, продолжает работать
5) Нагрузочный тест: Поведение под нагрузкой

# 1. Unit тест

**Unit-тест (модульный тест)** - это фрагмент кода, который:
- Вызывает одну конкретную единицу кода (функцию, метод, класс) — unit
- Подает на вход заранее известные данные
- Сравнивает результат с заранее известным ожидаемым значением
- Автоматически сообщает: PASSED (тест пройден) или FAILED (тест провален)

**Unit-тест = вход + вызов + ожидаемый выход + сравнение**

```python
# В проекте есть функция
def calculate_discount_price(price: float, discount: float) -> float:
    """Функция по расчету цены за минусом скидки"""
    return price * (1 - discount)
```

```python
calculate_discount_price(100, 0.2)
```

```python
# Основой большинства тестов является assert
def test_discount_price():
    assert calculate_discount_price(100, 0.2) == 80.0
```

```python
test_discount_price()
```

```python
# Кто-то ее поменял
def calculate_discount_price(price: float, discount: float) -> float:
    return price * discount
```

```python
calculate_discount_price(100, 0.2)
```

```python
test_discount_price()
```

# 2. Правило создания Unit тестов

```python
# Предположим, процент скидки зависит от числа заказов и должен быть отдельно рассчитан
def calculate_discount(order_count: int) -> float:
    """Возвращает скидку клиента в зависимости от числа заказов."""
    if order_count >= 10:
        return 0.15
    elif order_count >= 5:
        return 0.10
    else:
        return 0.0
```

```python
# Добавлять несколько тестов в одну функцию - плохо
def test_calculate_discount_manual():
    assert calculate_discount(0) == 0.0
    assert calculate_discount(5) == 0.10
    assert calculate_discount(10) == 0.15
    print("Все проверки пройдены!")
```

```python
test_calculate_discount_manual()
```

**Хороший тест отвечает на конкретный вопрос**

```python
def test_no_discount_for_new_customer():
    assert calculate_discount(0) == 0.0

def test_ten_percent_threshold():
    assert calculate_discount(5) == 0.10

def test_fifteen_percent_threshold():
    assert calculate_discount(10) == 0.15
```

```python
test_no_discount_for_new_customer()
```

```python
test_ten_percent_threshold()
```

```python
test_fifteen_percent_threshold()
```

# 3. Arrange — Act — Assert

1) Arrange (подготовка) - создать входные данные
2) Act (действие) - вызвать тестируемую функцию
3) Assert (проверка) - сравнить результат с ожидаемым

```python
def test_discount_price():
    # Arrange
    price = 100
    discount = 0.2

    # Act
    result = calculate_discount_price(price, discount)

    # Assert
    assert result == 80
```

```python
test_discount_price()
```

# 7. Параметризация тестов

```python
def add(a, b):
    return a + b
```

```python
@pytest.mark.parametrize(
    "a,b,expected",
    [
        (1, 2, 3),
        (10, 20, 30),
        (-1, 1, 0),
    ]
)
def test_add(a, b, expected):
    assert add(a, b) == expected
```

```python
ipytest.run()
```

```python
# Предположим, процент скидки зависит от числа заказов и должен быть отдельно рассчитан
def calculate_discount(order_count: int) -> float:
    """Возвращает скидку клиента в зависимости от числа заказов."""
    if order_count >= 10:
        return 0.15
    elif order_count >= 5:
        return 0.10
    else:
        return 0.0
```

```python
%%ipytest

@pytest.mark.parametrize(
    "order_count,expected",
    [
        (0, 0),
        (3, 0),
        (5, 0.1),
        (7, 0.1),
        (10, 0.15),
        (15, 0.15),
    ]
)
def test_discount(order_count, expected):
    assert calculate_discount(order_count) == expected
```

```python
ipytest.run()
```

# 8. Фикстуры

**Фикстура** - это функция-«поставщик» данных или ресурсов для тестов

Pytest сам передает результат fixture в тест как аргумент - по имени

```python
%%ipytest

@pytest.fixture
def numbers():
    return [10, 20, 30]

def test_sum(numbers):
    assert sum(numbers) == 60
```

```python
ipytest.run()
```

Представим, что мы работаем с датафреймом и получаем оттуда атрибуты для ML

```python
def average_salary(df):
    return df["salary"].mean()

def average_age(df):
    return df["age"].mean()

def fill_missing_values(df):
    return df.fillna(0)
```

```python
import pandas as pd
```

```python
%%ipytest

@pytest.fixture
def sample_data():
    return pd.DataFrame({"age": [20, 30, 40, np.nan], "salary": [100000, 200000, np.nan, 300000]})

def test_average_salary(sample_data):
    assert average_salary(sample_data) == 200000.0

def test_average_age(sample_data):
    assert average_age(sample_data) == 30.0

def test_fill_missing_values(sample_data):
    sample_data = fill_missing_values(sample_data)
    assert (sample_data.loc[3, "age"] == 0)&(sample_data.loc[2, "salary"] == 0)
```

# 9. Mock

```python
def get_vip_customers(connection):
    cursor = connection.cursor()
    cursor.execute("""
        SELECT c.customer_id, c.first_name, SUM(p.amount) AS total
        FROM customer c
        JOIN payment p ON c.customer_id = p.customer_id
        GROUP BY c.customer_id, c.first_name
        HAVING SUM(p.amount) > 150
    """)
    return cursor.fetchall()
```

```python
from unittest.mock import Mock, MagicMock
```

```python
fake_connection = Mock()
fake_connection.cursor.return_value.fetchall.return_value = [
    (1, "Иван", 200.50),
    (2, "Мария", 175.25),
]
result = get_vip_customers(fake_connection)
print(result)
```

```python
# Проверим, что курсор вернул все данные и больше ничего не содержит
fake_connection.cursor.assert_called_once()
```
