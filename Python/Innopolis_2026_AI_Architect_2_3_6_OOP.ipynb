**Инкапсуляция** - это объединение данных и методов работы с ними в одном объекте, с сокрытием внутренней реализации и контролем доступа к данным извне.

## 4.1. Соглашения об уровнях доступа в Python

- Public (публичный): имена без подчеркиваний (name). Доступны везде внутри и вне класса.

- Protected (защищенный): имя с одним подчеркиванием (_name). Сигнал для разработчиков, что атрибут предназначен только для внутреннего использования внутри класса и его наследников. Технически снаружи к нему все равно можно обратиться.

- Private (приватный): имя с двумя подчеркиваниями (__name). Запускает механизм mangling (искажения имени), превращая атрибут в _ClassName__name, что затрудняет случайный доступ извне, но не блокирует его полностью.

### 4.1.1. Public attribute

```python
class Customer:
    def __init__(self, email, first_name):
        self.email = email            # публичный
        self.first_name = first_name  # публичный
```

```python
c1 = Customer("anna@example.com", "Anna")
```

```python
print(c1.email)          # свободный доступ на чтение
```

```python
print(c1.first_name)
```

```python
c1.__dict__
```

```python
c1.first_name = "Anna Petrovna"   # свободный доступ на запись
```

```python
print(c1.first_name)
```

- Python не задействует никаких скрытых механизмов
- обычный атрибут объекта, хранящийся в object.\_\_dict\_\_
- используется по умолчанию для данных, изменение которых не требует контроля и не может нарушить целостность объекта
- большинство атрибутов в простых учебных примерах — публичные
- инкапсуляция нужна не везде, а там, где есть риск некорректного состояния

### 4.1.2. Protected attribute

```python
class Rental:
    def __init__(self, rental_id, film_title, rental_rate):
        self.rental_id = rental_id
        self.film_title = film_title
        self.rental_rate = rental_rate
        self._late_fee_accrued = 0.0    # защищенный атрибут для внутренних расчетов класса

    def accrue_fee(self, amount):
        self._late_fee_accrued += amount

    def get_total_fee(self):
        return round(self.rental_rate + self._late_fee_accrued, 2)
```

- это соглашение, а не механизм защиты
- интерпретатор Python никак не проверяет и не блокирует доступ к _attribute
- при from module import * имена с подчеркиванием не импортируются автоматически
- используется, когда атрибут нужен внутри класса и, возможно, в дочерних классах (при наследовании), но не предназначен для внешнего кода

## 4.2. Геттеры и сеттеры

- **Геттер** (getter) - метод, который возвращает значение атрибута
- **Сеттер** (setter) - метод, который изменяет значение атрибута, обычно с проверкой корректности нового значения
- Проблема, которую они решают: если разрешить прямое изменение приватного атрибута напрямую, никакой контроль невозможен, а если полностью запретить доступ, то объектом станет неудобно пользоваться

### 4.2.1. Методы get() и set()

```python
class Rental:
    def __init__(self, rental_id, film_title):
        self.rental_id = rental_id
        self.film_title = film_title
        self.__status = "rented"

    def get_status(self):
        return self.__status

    def set_status(self, new_status):
        allowed = ["rented", "returned", "lost"]
        if new_status not in allowed:
            raise ValueError(f"Недопустимый статус: {new_status}")
        self.__status = new_status
```

```python
r1 = Rental(5001, "ACADEMY DINOSAUR")
```

```python
r1.__dict__
```

```python
print(r1.get_status())
```

```python
r1.set_status("returned")
```

```python
print(r1.get_status())
```

```python
r1.set_status("stolen")
```

- Недостаток такого подхода в Python: он утяжеляет код
- Каждое обращение к статусу требует вызова метода со скобками — r1.get_status() вместо естественного r1.status
- Это не "питонический" вариант решения проблемы

### 4.2.2. Декоратор @property

```python
class Rental:
    def __init__(self, rental_id, film_title):
        self.rental_id = rental_id
        self.film_title = film_title
        self.__status = "rented"

    @property
    def status(self):          # это геттер, оформленный как @property
        print("вызван геттер status")
        return self.__status
```

```python
r1 = Rental(5001, "ACADEMY DINOSAUR")
```

```python
# Метод status, оформленный декоратором, вызывается без круглых скобок
print(r1.status)
```

```python
# Попробуем вызвать метод status с круглыми скобками
r1.status()
```

```python
# Попробуем поменять значение атрибута status на returned
r1.status = "returned" # сеттера пока нет
```

- r1.status выглядит как обращение к обычному атрибуту
- на самом деле каждый раз выполняется код метода status()
- это позволяет добавить логику (проверку, вычисление, логирование) без изменения интерфейса
- весь код, использующий r1.status, продолжит работать без переделки, даже если внутри появится сложная логика

### 4.2.3. Декоратор @x.setter

```python
class Rental:
    VALID_TRANSITIONS = {
        "rented": ["returned", "lost"],
        "returned": list(),
        "lost": list(),
    }

    def __init__(self, rental_id, film_title):
        self.rental_id = rental_id
        self.film_title = film_title
        self.__status = "rented"

    @property
    def status(self):
        return self.__status

    @status.setter
    def status(self, new_status):
        if new_status not in self.VALID_TRANSITIONS[self.__status]:
            raise ValueError(f"Нельзя перейти из '{self.__status}' в '{new_status}'")
        print(f"Статус меняется: {self.__status} -> {new_status}")
        self.__status = new_status
```

```python
r1 = Rental(5001, "ACADEMY DINOSAUR")
```

```python
print(r1.__dict__)
```

```python
r1.status = "returned"      # выглядит как обычное присваивание, но вызывается сеттер
```

```python
print(r1.status)            # работает декоратор @property как геттер
```

```python
r1.status = "rented"
```

Еще пример использования декораторов для присваивания и получения атрибутов

```python
class Customer:
    def __init__(self, first_name, last_name, birth_year):
        self.first_name = first_name
        self.last_name = last_name
        self.birth_year = birth_year

    @property
    def full_name(self):                 # вычисляемое свойство — нет self.__full_name!
        return f"{self.first_name} {self.last_name}"

    @property
    def age(self):
        from datetime import date
        return date.today().year - self.birth_year
```

```python
c1 = Customer("Anna", "Petrova", 1994)
```

```python
print(c1.full_name)   # вычислено на лету из двух атрибутов
```

```python
print(c1.age)         # вычислено из текущего года, нигде не хранится напрямую
```

### 4.2.4. Декоратор @x.deleter

```python
class Rental:
    def __init__(self, rental_id, film_title):
        self.rental_id = rental_id
        self.__film_title = film_title

    @property
    def film_title(self):
        return self.__film_title

    @film_title.setter
    def film_title(self, value):
        self.__film_title = value

    @film_title.deleter
    def film_title(self):
        print(f"Запись о фильме '{self.__film_title}' удалена из аренды {self.rental_id}")
        self.__film_title = None
```

## 4.3. Пример правильной работы инкапсуляции

На старте проекта задача простая: у аренды есть срок в днях. 

Никакой валидации не требуется — пишем как есть.

```python
class Rental:
    def __init__(self, rental_id, film_title, rental_days):
        self.rental_id = rental_id
        self.film_title = film_title
        self.rental_days = rental_days
```

```python
r1 = Rental(5001, "ACADEMY DINOSAUR", 7)
```

```python
# Посмотрим значение атрибута
print(r1.rental_days)
```

```python
# Изменим значение срока проката на 14 дней
r1.rental_days = 14
```

```python
# Проверим, что срок проката изменился
print(r1.rental_days)
```

Через некоторое время в проде выясняется: кто-то присвоил r1.rental_days = -3

Это сломало расчет даты возврата.

Нужно запретить некорректные значения, не меняя способ обращения снаружи — иначе придется переписывать весь код, использующий rental.rental_days.

Решение: приватный атрибут-хранилище __rental_days + @property с тем же именем rental_days, что и было раньше.

```python
class Rental:
    def __init__(self, rental_id, film_title, rental_days):
        self.rental_id = rental_id
        self.film_title = film_title
        self.rental_days = rental_days

    @property
    def rental_days(self):
        return self.__rental_days

    @rental_days.setter
    def rental_days(self, value):
        if not isinstance(value, int) or value <= 0:
            raise ValueError("Срок аренды должен быть положительным целым числом")
        self.__rental_days = value
```

```python
r1 = Rental(5001, "ACADEMY DINOSAUR", 7)
```

```python
# Посмотрим значение атрибута
print(r1.rental_days)
```

```python
# Изменим значение срока проката на 14 дней
r1.rental_days = 14
```

```python
# Проверим, что срок проката изменился
print(r1.rental_days)
```

```python
r1.rental_days = -3
```

```python
print(r1.__dict__)
```

Даже строка self.rental_days = rental_days внутри \_\_init\_\_ теперь фактически вызывает сеттер

Валидация срабатывает и при создании объекта, и мы не написали для этого отдельного кода.

Еще через время архитектурное решение: срок аренды в базе данных теперь хранится не в днях, а как разница return_date - rental_date в часах

Так удобнее для аналитики просрочек

Внутреннее представление меняется полностью, но снаружи код по-прежнему должен работать со сроком в днях

```python
class Rental:
    def __init__(self, rental_id, film_title, rental_days):
        self.rental_id = rental_id
        self.film_title = film_title
        self.rental_days = rental_days

    @property
    def rental_days(self):
        return self.__rental_hours // 24        # реализация теперь другая: пересчет из часов

    @rental_days.setter
    def rental_days(self, value):
        if not isinstance(value, int) or value <= 0:
            raise ValueError("Срок аренды должен быть положительным целым числом")
        self.__rental_hours = value * 24          # хранится в часах, а не в днях!
```

```python
r1 = Rental(5001, "ACADEMY DINOSAUR", 7)
```

```python
# Посмотрим значение атрибута
print(r1.rental_days)
```

```python
# Изменим значение срока проката на 14 дней
r1.rental_days = 14
```

```python
# Проверим, что срок проката изменился
print(r1.rental_days)
```

```python
print(r1.__dict__)
```

# 5. Наследование

**Наследование** - механизм, позволяющий создать новый класс (потомок/дочерний), который автоматически получает атрибуты и методы существующего класса (родителя/базового), и может их дополнять или переопределять.

```python
class Rental:
    def __init__(self, rental_id, film_title):
        self.rental_id = rental_id
        self.film_title = film_title
        self.status = "rented"

    def return_film(self):
        self.status = "returned"
        print(f"Аренда {self.rental_id} завершена, диск возвращен")
```

```python
class OverdueRental(Rental):
    pass
```

```python
o1 = OverdueRental(6001, "TORQUE BOUND")
```

```python
print(o1.rental_id)  # Унаследованный атрибут
```

```python
o1.return_film()     # Унаследованный метод
```

Чтобы добавить новые поля, но не переписывать логику родителя заново, вызываем конструктор родителя через super():

```python
class OverdueRental(Rental):
    def __init__(self, rental_id, film_title, days_overdue, late_fee_per_day):
        super().__init__(rental_id, film_title)   # вызов __init__ родителя
        self.days_overdue = days_overdue
        self.late_fee_per_day = late_fee_per_day

    def calculate_fee(self):
        return round(self.days_overdue * self.late_fee_per_day, 2)
```

```python
o1 = OverdueRental(
    6001, 
    "TORQUE BOUND",
    days_overdue=3, 
    late_fee_per_day=1.5
)
```

```python
print(o1.calculate_fee())
```

```python
o1.return_film()      # унаследованный метод
```

Переопределение методов (method overriding): дочерний класс может заменить поведение метода родителя, определив метод с тем же именем

```python
class OverdueRental(Rental):
    def __init__(self, rental_id, film_title, days_overdue, late_fee_per_day):
        super().__init__(rental_id, film_title)
        self.days_overdue = days_overdue
        self.late_fee_per_day = late_fee_per_day

    def calculate_fee(self):
        return round(self.days_overdue * self.late_fee_per_day, 2)

    def return_film(self):                     # переопределяем return_film()
        fee = self.calculate_fee()
        print(f"Начислен штраф: {fee}$ за {self.days_overdue} дней просрочки")
        super().return_film()                    # но все равно используем логику родителя
```

```python
o1 = OverdueRental(6001, "TORQUE BOUND", 3, 1.5)
```

```python
o1.return_film()
```

# 6. Полиморфизм

**Полиморфизм**- способность объектов разных классов отвечать на один и тот же вызов метода каждый своим, специфичным для класса, образом, без необходимости знать заранее, объект какого именно класса перед нами

```python
all_rentals = [
    Rental(5001, "ACADEMY DINOSAUR"),
    OverdueRental(6001, "TORQUE BOUND", 3, 1.5),
    Rental(5002, "AIRPORT POLLOCK"),
]
```

```python
# Плохо: приходится проверять тип вручную
for rental in all_rentals:
    if isinstance(rental, OverdueRental):
        fee = rental.calculate_fee()
        rental.return_film()
    else:
        rental.return_film()
```

```python
# один и тот же вызов, разное поведение!
for rental in all_rentals:
    rental.return_film()
```

В Python полиморфизм не требует общего родителя. Если объект «умеет» то, что от него ожидают — этого достаточно

```python
class EmailNotification:
    def send(self, message):
        print(f"[EMAIL] Отправлено: {message}")

class SmsNotification:
    def send(self, message):
        print(f"[SMS] Отправлено: {message}")

class PushNotification:
    def send(self, message):
        print(f"[PUSH] Отправлено: {message}")
```

```python
channels = [EmailNotification(), SmsNotification(), PushNotification()]

for channel in channels:
    channel.send("Напоминаем: диск 'TORQUE BOUND' просрочен на 3 дня")
```
