# Декоратор 

Декоратор - это функция, которая принимает другую функцию и возвращает новую функцию, добавляя ей какое-то новое поведение

## Создадим декоратор вручную

```python
def decorator(func):
    def wrapper():
        print("Начало работы")
        func()
        print("Конец работы")
    return wrapper

def hello():
    print("Hello!")
```

```python
# Запустим функцию без декоратора
hello()
```

```python
# Передаем функцию hello как аргумент декоратора и результат присваемаем переменной hello
hello = decorator(hello)
```

```python
hello()
```

## Синтаксис @

Вернем изначальную функцию

```python
def decorator(func):
    def wrapper():
        print("Начало работы")
        func()
        print("Конец работы")
    return wrapper

def hello():
    print("Hello!")
```

Вызовем

```python
hello()
```

Теперь используем **декораторную запись**

```python
@decorator
def hello():
    print("Hello!")
```

```python
hello()
```

Принцип работы большинства декораторов:

wrapper:

-- Действия до

|

-- Основная функция

|

-- Действия после

**Попробуем декоратор с другой функцией**

```python
@decorator
def countdown():
    i=3
    while i>0:
        print(f"=={i}==")
        i-=1
    print("Пуск!")
```

```python
countdown()
```

# Декораторы с аргументами 

## Декоратор для функции с аргументами

Оригинальная функция с аргументом

```python
def decorator(func):
    def wrapper():
        print("Начало работы")
        func()
        print("Конец работы")
    return wrapper

def hello(name):
    print(f"Hello, {name}!")
```

```python
hello("Denis")
```

Пробуем добавить декоратор

```python
@decorator
def hello(name):
    print(f"Hello, {name}!")
```

```python
hello("Denis")
```

wrapper() не принимает аргументы!

## Создаем wrapper с аргументами

```python
def decorator(func):
    def wrapper(*args, **kwargs):
        print("Начало работы")
        result = func(*args, **kwargs)
        print("Конец работы")
        return result
    return wrapper
```

```python
@decorator
def hello(name):
    print(f"Hello, {name}!")
```

```python
hello("Denis")
```

```python
@decorator
def add(a, b):
    return a + b
```

```python
x = add(10, 15)
print(x)
```

## functools.wraps

```python
def hello(name):
    """Это документация функции"""
    print(f"Hello, {name}!")
```

```python
# Вызываем документацию по функции hello
help(hello)
```

```python
def decorator(func):
    def wrapper(*args, **kwargs):
        print("Начало работы")
        result = func(*args, **kwargs)
        print("Конец работы")
        return result
    return wrapper

@decorator
def hello(name):
    """Это документация функции"""
    print(f"Hello, {name}!")
```

```python
help(hello)
```

Python присвоил объект wrapper функции hello и документация потерялась!

```python
from functools import wraps
```

```python
def decorator(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        print("Начало работы")
        result = func(*args, **kwargs)
        print("Конец работы")
        return result
    return wrapper
```

```python
@decorator
def hello(name):
    """Это документация функции"""
    print(f"Hello, {name}!")
```

```python
help(hello)
```

Теперь документация сохранилась!

# Примеры практических декораторов

## Измерение времени

```python
from functools import wraps
import time
```

```python
def timer(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        start = time.perf_counter()
        result = func(*args, **kwargs)
        end = time.perf_counter()
        print(
            f"{func.__name__}: "
            f"{end - start:.4f} сек."
        )
        return result
    return wrapper
```

```python
@timer
def train_model():
    time.sleep(3)
    return "model"
```

```python
model = train_model()
```

## Кэширование

```python
from functools import lru_cache

@lru_cache(maxsize=128)
def fibonacci(n):
    time.sleep(0.1)
    if n < 2:
        return n
    else:
        return fibonacci(n-1) + fibonacci(n-2)
```

```python
# Вызываем функцию первый раз
start_time = time.time()
result = fibonacci(100)
#print(result)
end_time = time.time()
print(f"Время работы: {end_time - start_time}")
```

```python
# Вызываем функцию в следующий
start_time = time.time()
result = fibonacci(100)
#print(result)
end_time = time.time()
print(f"Время работы: {end_time - start_time}")
```

# Контекстный менеджер

## Контекст

```python
file = open("data.csv")
data = file.read()
file.close()
```

Но если в data случится ошибка, то file.close() не выполнится

```python
with open("data.csv") as file:
    data = file.read()
```

После выхода из with файл автоматически закрывается, даже если внутри произошла ошибка

Идея:

1) Что сделать перед входом в блок
2) Блок основного кода
3) Что сделать при выходе

## Принцип работы 

```python
class MyContext:
    def __enter__(self):
        print("Вход в контекст")
        return self
    def __exit__(self, exc_type, exc_value, traceback):
        print("Выход из контекста")
```

```python
with MyContext():
    print("Работаем")
```

exc_type - ошибка

exc_value - информация об ошибке

```python
class MyContext:
    def __enter__(self):
        print("Вход в контекст")
        return self
    def __exit__(self, exc_type, exc_value, traceback):
        print("Выход из контекста")
        if exc_type:
            print("Произошла ошибка:", exc_value)
```

```python
with MyContext():
    10/0
```

## contextlib

```python
from contextlib import contextmanager
```

```python
@contextmanager
def my_context():
    print("Вход в контекст")
    yield
    print("Выход из контекста")
```

```python
with my_context():
    print("Работаем")
```

Все до yield выполняется при входе

Все после yield выполняется при выходе

```python
with my_context():
    10/0
```

Если случилась ошибка, то часть после yield не выполняется

```python
@contextmanager
def my_context():
    print("Вход в контекст")
    try: 
        yield
    finally:
        print("Выход из контекста")
```

```python
with my_context():
    10/0
```

# Примеры контекстных менеджеров

## Измерение времени

```python
from contextlib import contextmanager
import time
```

```python
@contextmanager
def timer():
    start = time.perf_counter()
    try:
        yield
    finally:
        end = time.perf_counter()
        print(f"Время: {end - start:.4f} сек.")
```

```python
def train_model():
    time.sleep(3)
    return "model"
```

```python
with timer():
    train_model()
```
