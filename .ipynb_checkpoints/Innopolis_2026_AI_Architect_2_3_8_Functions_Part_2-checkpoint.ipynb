# 1. Локальные переменные

Переменная, созданная внутри функции, обычно доступна только внутри этой функции

````python
def example():
    x = 10
    print(x)
````

````python
example()
````

````python
print(x)
````

x существует только внутри функции. Такая переменная называется локальной переменной.

# 2. Глобальные переменные

Переменная, созданная вне функций, обычно находится в глобальной области видимости

````python
def show():
    print(x)
````

````python
x = 10
show()
````

Функция может прочитать глобальную переменную.

Однако изменять глобальные переменные внутри функции без необходимости не рекомендуется.

````python
def change():
    x = 20
````

````python
x = 10
change()
print(x)
````

# 3. Оператор global

Если необходимо изменить глобальную переменную внутри функции

````python
def change():
    global x
    x = 20
````

````python
x = 10
change()
print(x)
````

Использование global считается плохой практикой

Лучше передавать данные в функцию и возвращать результат

````python
def change(x):
    return 20
````

````python
x = change(x)
print(x)
````

# 4. Передача объектов в функцию

При передаче аргумента функция работает с объектом, на который ссылается переменная

````python
def add_item(items):
    items.append("Новый элемент")
````

````python
my_list = ["A", "B"]
add_item(my_list)
print(my_list)
````

Список изменился

**Неизменяемые объекты**

````python
def change(x):
    x = x + 1
````

````python
number = 10
change(number)
print(number)
````

**Изменяемые объекты**

````python
def add_item(data):
    data.append(100)
````

````python
numbers = [1, 2, 3]
add_item(numbers)
print(numbers)
````

**Изменяемые объекты и значения по умолчанию**

````python
def add_item(item, items=[]):
    items.append(item)
    return items
````

````python
add_item("A")
````

````python
add_item("B")
````

Следует использовать:

````python
def add_item(item, items=None):
    if items is None:
        items = []
    items.append(item)
    return items
````

````python
add_item("A")
````

````python
add_item("B")
````

````python
add_item("B", items = ["A"])
````

**Пример неизменяемости внешней переменнной**

````python
# Неправильно: список изменяется
def add_tax(orders):
    for i in range(len(orders)):
        orders[i] *= 1.2   # меняет исходный список!
    return orders
````

````python
orders = [1200, 500, 800]
with_tax = add_tax(orders)
print(orders)
print(with_tax)
````

````python
# Правильно: список не изменяется
def add_tax(orders):
    return [amount * 1.2 for amount in orders]  # новый список
````

````python
orders = [1200, 500, 800]
with_tax = add_tax(orders)
print(orders)
print(with_tax)
````

# 5. Чистые функции

**Детерминированность**

````python
# Чистая
def add(a, b):
    return a + b
````

````python
print(add(2,3))
print(add(2,3))
print(add(2,3))
````

````python
# Нечистая
import random
def roll():
    return random.randint(1, 6)
````

````python
print(roll())
print(roll())
print(roll())
````

````python
# Нечистая
from datetime import datetime
def today():
    return datetime.now()
````

````python
print(today())
print(today())
print(today())
````

**Отсутствие побочных эффектов**

````python
# Чистая
def append(lst, item):
    return lst + [item]
````

````python
lst = [1,2,3]
append(lst, 4)
````

````python
lst
````

````python
# Нечистая
def append_bad(lst, item):
    lst.append(item)
    return lst
````

````python
lst = [1,2,3]
append_bad(lst, 4)
````

````python
lst
````

# 6. Специальные случаи

**Функции могут вызывать другие функции**

````python
def calculate_sum(a, b):
    return a + b
````

````python
def calculate_average(a, b):
    total = calculate_sum(a, b)
    return total / 2
````

````python
result = calculate_average(10, 20)
print(result)
````

**Функция может вызывать саму себя: рекурсия**

````python
def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n - 1)
````

````python
factorial(5)
````

Логика:

```python
5 * factorial(4)
4 * factorial(3)
3 * factorial(2)
2 * factorial(1)
```

Очень важно иметь условие остановки:
```python
if n == 1:
    return 1
```

Без него функция будет вызывать себя бесконечно, пока Python не достигнет ограничения глубины рекурсии.

````python
def factorial(n):
    return n * factorial(n - 1)
````

````python
factorial(5)
````

**Вложенные функции**

````python
def outer():
    x = 10
    def inner():
        return x        # inner «помнит» x — это замыкание
    return inner
````

````python
f = outer()
````

````python
f()
````

**Области видимости**

````python
x = "global"
def outer():
    x = "enclosing"
    def inner():
        x = "local"
    inner()
    print(x)
````

````python
outer()
````

````python
x = "global"
def outer():
    x = "enclosing"
    def inner():
        nonlocal x
        x = "local"
    inner()
    print(x)
````

````python
outer()
````

# 7. Функции как объекты

В Python функции являются объектами. Их можно присваивать переменным:

````python
def greet():
    print("Привет!")
````

````python
my_function = greet
````

````python
my_function()
````

**Передача функции как аргумента**

````python
def apply_function(function, value):
    return function(value)

def square(x):
    return x ** 2

def cube(x):
    return x ** 3
````

````python
apply_function(square, 5)
````

````python
apply_function(cube, 5)
````

**Лямбда-функции**

````python
def square(x):
    return x ** 2
````

````python
square = lambda x: x ** 2
````

````python
print(square(5))
````

````python
numbers = [1, 2, 3, 4, 5]
result = list(map(lambda x: x ** 2, numbers))
print(result)
````

# 8. Функции высших порядков

**map**

````python
def square(x):
    return x ** 2
````

````python
numbers = [1, 2, 3, 4]
result = map(square, numbers)
print(result)
print(list(result))
````

````python
# Эквивалент
[x ** 2 for x in numbers]
````

**filter**

````python
def is_even(x):
    return x % 2 == 0
````

````python
numbers = [1, 2, 3, 4, 5, 6]
result = filter(is_even, numbers)
print(result)
print(list(result))
````

````python
# Эквивалент
[x for x in numbers if x % 2 == 0]
````

**reduce**

````python
from functools import reduce

numbers = [1, 2, 3, 4]
reduce(lambda acc, x: acc + x, numbers)
````

````python
reduce(lambda acc, x: acc + x, numbers, 1)
````

# 9. Функции и обработка ошибок

Функция может содержать обработку исключений.

````python
def divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("На ноль делить нельзя")
````

````python
result = divide(10, 0)
````

````python
def divide(a, b):
    if b == 0:
        return None
    return a / b
````

````python
result = divide(10, 0)
print(result)
````

````python
def divide(a, b):
    if b == 0:
        raise ValueError("Делитель не может быть равен нулю")
    return a / b
````

````python
result = divide(10, 0)
print(result)
````
