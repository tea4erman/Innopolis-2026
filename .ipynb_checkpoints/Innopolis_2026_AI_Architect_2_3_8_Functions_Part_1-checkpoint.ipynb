# 1. Создание функции

**Функция** - это именованный блок кода, который выполняет определенную задачу

Например, вместо того чтобы несколько раз писать:

````python
print("Добро пожаловать!")
````

можно создать функцию:

````python
def greet():
    print("Добро пожаловать!")
````

Теперь функцию можно вызвать:

````python
greet()
````

Для создания функции используется ключевое слово **def**

Общий синтаксис:

```python
def имя_функции(параметры):

    тело_функции
```

Например:

````python
def hello():
    print("Привет!")
````

Здесь:

- `def` — ключевое слово для создания функции;
- `hello` — имя функции;
- `()` — место для параметров;
- `:` — обозначает начало тела функции;
 код внутри функции должен иметь отступ.

Вызов функции:

````python
hello()
````

# 4. Позиционные и именованные аргументы

**Позиционные аргументы**

При обычной передаче аргументы связываются с параметрами по порядку:

````python
def show_person(name, age):
    print(f"Имя: {name}")
    print(f"Возраст: {age}")
````

````python
show_person("Анна", 30)
````

То есть:

```text
name = "Анна"
age = 30
```

Если поменять аргументы местами:

````python
show_person(30, "Анна")
````

результат будет неправильным с точки зрения логики программы

**Именованные аргументы**

Можно передавать аргументы, явно указывая имя параметра

````python
show_person(name = "Анна", age = 30)
````

````python
show_person(age = 30, name = "Анна")
````

Несмотря на изменение порядка, Python понимает, какое значение относится к какому параметру

# 6. Произвольное количество аргументов

**Неименованные аргументы**

Для передачи произвольного количесчтва аргументов используется *args.

````python
def calculate_sum(*args):
    return sum(args)
````

````python
calculate_sum(1, 2)
````

````python
calculate_sum(1, 2, 3, 4, 5)
````

````python
calculate_sum(10, 20, 30)
````

Внутри функции args является кортежем

````python
def show(*args):
    print(args)
````

````python
show(1, 2, 3)
````

````python
def example(*values):
    print(values)
````

````python
example(1,2,3,4)
````

**Именованные аргументы**

````python
def show_info(**kwargs):
    print(kwargs)
````

````python
show_info(
    name="Иван",
    age=25,
    city="Москва"
)
````

````python
def show_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")
````

````python
show_info(
    name="Иван",
    age=25,
    city="Москва"
)
````

**Все аргументы вместе**

Общий порядок обычно выглядит так:
- обычные_параметры,
- *args,
- параметры_после_star,
- **kwargs

````python
def example(a, b = 10, *args, **kwargs):
    print(a)
    print(b)
    print(args)
    print(kwargs)
````

````python
example(1, 2, 3, 4, name="Иван", age=25)
````

# 8. Оператор return

Функция может не только выполнять действия, но и возвращать результат. Для этого используется return.

````python
def add(a, b):
    return a + b
````

````python
result = add(10, 20)
print(result)
````

print() и return() - это не одно и то же

````python
def calculate():
    print(10 + 20)
````

````python
x = calculate()
print(x)
````

Функция вывела число 30, но ничего не вернула.
    
Если функция не содержит return, Python автоматически возвращает None

После выполнения return функция сразу завершает работу

````python
def example():
    print("Начало")
    return 10
    print("Конец")
````

````python
example()
````

```python
print("Конец")
```

никогда не будет выполнена.


Python позволяет вернуть несколько значений

````python
def calculate(a, b):
    total = a + b
    difference = a - b
    return total, difference
````

````python
result = calculate(10, 5)
print(result)
````

````python
type(result)
````

````python
total, difference = calculate(10, 5)
print(total)
print(difference)
````

# 9. Docstring

# 11. Когда следует выделять код в функции

## 11.1. Если код повторяется несколько раз (Принцип DRY - Don't Repeat Yourself)

Признак того, что пора выделять функцию: вы копируете блок кода через Ctrl+C/Ctrl+V и меняете в нем одну-две переменные

````python
# Плохо
order1_total = 1200 * 1.2 if 1200 > 1000 else 1200 * 1.1
order2_total = 500 * 1.2 if 500 > 1000 else 500 * 1.1
order3_total = 800 * 1.2 if 800 > 1000 else 800 * 1.1
print(order1_total)
print(order2_total)
print(order3_total)
````

````python
# Хорошо
def calculate_total_with_tax(amount):
    tax_rate = 1.2 if amount > 1000 else 1.1
    return amount * tax_rate
````

````python
order1_total = calculate_total_with_tax(1200)
order2_total = calculate_total_with_tax(500)
order3_total = calculate_total_with_tax(800)
print(order1_total)
print(order2_total)
print(order3_total)
````

## 11.2. Декомпозиция: разбиение сложной задачи на именованные шаги

````python
# Плохо: все действия в одной функции
def handle_ticket(ticket):
    text = ticket["text"].strip().lower()
    text = text.replace("!", "").replace("?", "")
    if "возврат" in text or "верните деньги" in text:
        category = "refund"
    elif "не пришел" in text or "не доставлен" in text:
        category = "delivery"
    else:
        category = "general"
    priority = 1 if ticket["customer_tier"] == "VIP" else 2
    if category == "refund":
        priority = 1
    return {"category": category, "priority": priority}
````

````python
ticket = {"text": "я хочу оформить возврат", "customer_tier" : "VIP"}
handle_ticket(ticket)
````

````python
# Хорошо: все действия разнесены по разным частям
def normalize_text(text):
    return text.strip().lower().replace("!", "").replace("?", "")

def detect_category(text):
    if "возврат" in text or "верните деньги" in text:
        return "refund"
    if "не пришел" in text or "не доставлен" in text:
        return "delivery"
    return "general"

def calculate_priority(customer_tier, category):
    if customer_tier == "VIP" or category == "refund":
        return 1
    return 2

def handle_ticket(ticket):
    text = normalize_text(ticket["text"])
    category = detect_category(text)
    priority = calculate_priority(ticket["customer_tier"], category)
    return {"category": category, "priority": priority}
````

````python
ticket = {"text": "я хочу оформить возврат", "customer_tier" : "VIP"}
````

````python
text = normalize_text(ticket["text"])
print(text)
````

````python
category = detect_category(text)
print(text)
````

````python
priority = calculate_priority(ticket["customer_tier"], category)
print(priority)
````

````python
handle_ticket(ticket)
````

## 11.3. Переиспользование: одна логика — много точек вызова

Если бы завтра изменился формат ФИО (например, добавили отчество), без функции пришлось бы искать и чинить код в трёх разных местах.

````python
def format_customer_name(first_name, last_name):
    return f"{last_name.upper()} {first_name.capitalize()}"

# Используется в email-уведомлении
def build_email_greeting(customer):
    return f"Здравствуйте, {format_customer_name(customer['first'], customer['last'])}!"

# и в отчете для менеджера
def build_report_row(customer, ticket_count):
    return f"{format_customer_name(customer['first'], customer['last'])}: {ticket_count} тикетов"

# и в чат-боте
def build_chat_signature(customer):
    return f"— {format_customer_name(customer['first'], customer['last'])}, служба поддержки"
````

````python
customer = {"first" : "Иван", "last" : "Петров"}

email = build_email_greeting(customer)
report = build_report_row(customer, ticket_count = 5)
chat = build_chat_signature(customer)

print(email)
print(report)
print(chat)
````

## 11.4. Параметризация вместо зашитых значений (магических чисел)

Как только число или строка используется больше одного раза с разным смыслом - это сигнал вынести значение в параметр функции (или в константу).

````python
# Плохо: "магические" числа без объяснения смысла
def check_ticket(ticket):
    if ticket["hours_open"] > 24:
        return "SLA нарушен"
    return "в норме"
````

````python
check_ticket({"hours_open": 30})
````

````python
# Хорошо: параметр с говорящим именем + возможность переиспользовать с другим лимитом
def check_sla(ticket, sla_hours_limit):
    if ticket["hours_open"] > sla_hours_limit:
        return "SLA нарушен"
    return "в норме"
````

````python
print(check_sla({"hours_open": 30}, sla_hours_limit=24)) 
print(check_sla({"hours_open": 2}, sla_hours_limit=1))    
````

## 11.5. Функция как единица тестирования

````python
def calculate_refund_amount(order_amount, days_since_purchase, is_defective):
    if is_defective:
        return order_amount      # полный возврат за брак
    if days_since_purchase > 14:
        return 0                 # срок возврата истек
    return order_amount * 0.9    # комиссия за возврат 10%
````

````python
# Функцию легко проверить набором тестовых случаев, не поднимая базу данных и не создавая реальный заказ
assert calculate_refund_amount(1000, 5, False) == 900
assert calculate_refund_amount(1000, 20, False) == 0
assert calculate_refund_amount(1000, 20, True) == 1000
print("Все проверки пройдены")
````
